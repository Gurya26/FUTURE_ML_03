```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
data = pd.DataFrame({
    "resume": [
        "Python developer with machine learning and data analysis skills",
        "Java developer with spring boot and backend experience",
        "Data scientist with Python, deep learning, and NLP",
        "Frontend developer with HTML CSS JavaScript React",
        "Machine learning engineer with Python and AI experience"
    ]
})

data
```

```python
job_description = "Looking for a Python developer with machine learning and data science skills"
```

```python
data['resume'] = data['resume'].str.lower()
vectorizer = TfidfVectorizer()

resume_vectors = vectorizer.fit_transform(data['resume'])
job_vector = vectorizer.transform([job_description])
```

```python
similarity = cosine_similarity(resume_vectors, job_vector)
data['score'] = similarity

data
```

```python
ranked_data = data.sort_values(by='score', ascending=False)

print("Top Candidates:\n")
print(ranked_data)
```

```python
required_skills = ["python", "machine learning", "data science"]

def check_skills(resume):
    resume = resume.lower()
    matched = [skill for skill in required_skills if skill in resume]
    missing = [skill for skill in required_skills if skill not in resume]
    return matched, missing

print("\nSkill Analysis:\n")

for i, row in ranked_data.iterrows():
    matched, missing = check_skills(row['resume'])
    print(f"Resume: {row['resume']}")
    print("Matched Skills:", matched)
    print("Missing Skills:", missing)
    print("-"*50)
```

```python
print("\nInterpretation:")
print("Candidates with higher matching scores are more suitable for the job role")
```

```python
best = ranked_data.iloc[0]

print("\nBest Candidate:")
print("Resume:", best['resume'])
print("Score:", round(best['score'], 3))
```
